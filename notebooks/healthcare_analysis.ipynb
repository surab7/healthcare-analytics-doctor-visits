# Healthcare Analytics for Doctor Visits

Exploratory analysis of the supplied dataset. These comparisons are descriptive and do not prove causation.

```python
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt
DATA_PATH = Path('../data/doctor_visits.csv')
df = pd.read_csv(DATA_PATH)
df.head()
```

```python
print('Shape:', df.shape)
display(df.dtypes.to_frame('dtype'))
display(df.isna().sum().to_frame('missing_values'))
display(df['visits'].describe())
```

```python
zero_visits = (df['visits'] == 0).sum()
print(f'Zero visits: {zero_visits:,} / {len(df):,} ({zero_visits / len(df):.1%})')
display(df.groupby('illness')['visits'].agg(['count', 'mean', 'median']).round(3))
```

```python
df.groupby('illness')['visits'].mean().plot(kind='bar', rot=0)
plt.title('Average Doctor Visits by Illness Count')
plt.xlabel('Illness count')
plt.ylabel('Average visits')
plt.tight_layout()
plt.show()
```
